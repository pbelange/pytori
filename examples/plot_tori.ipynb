```python
import numpy as np
import pytori as pt

s1 = pt.BaseSeries({(1,0):np.pi+np.sqrt(2)*1j})/10
s2 = pt.FourierSeries({(1,0):np.pi+np.sqrt(2)*1j})/10

display(s1)
display(s2)


display(s1.conj())
display(s2.conj())
```
